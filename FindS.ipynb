```python
data = [
    ["Dell", "i5", "8GB", "SSD", "Good", "Yes"],
    ["HP", "i5", "8GB", "SSD", "Good", "Yes"],
    ["Lenovo", "i3", "4GB", "HDD", "Average", "No"],
    ["HP", "i5", "16GB", "SSD", "Good", "Yes"]
]

n = len(data[0]) - 1

S = ["Ø"] * n

for iteration, row in enumerate(data, 1):

    x = row[:-1]
    label = row[-1]

    print("\nIteration", iteration)
    print("Example:", x)
    print("Class:", label)

    if label == "Yes":

        for i in range(n):

            if S[i] == "Ø":
                S[i] = x[i]

            elif S[i] != x[i]:
                S[i] = "?"

    print("S =", S)

print("\nFinal Hypothesis:")
print(S)
```
