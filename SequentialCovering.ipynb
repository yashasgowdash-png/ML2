```python
data = [
    ["High", "Good", "Yes", "Low", "Yes"],
    ["High", "Good", "No", "Low", "Yes"],
    ["Medium", "Good", "Yes", "Medium", "Yes"],
    ["Low", "Good", "Yes", "Low", "Yes"],
    ["High", "Poor", "Yes", "High", "No"],
    ["Low", "Poor", "No", "High", "No"],
    ["Medium", "Poor", "Yes", "High", "No"],
    ["High", "Good", "Yes", "Medium", "Yes"],
    ["Medium", "Good", "No", "Low", "Yes"],
    ["Low", "Good", "No", "Medium", "No"],
    ["High", "Poor", "No", "Medium", "No"],
    ["Medium", "Good", "Yes", "Low", "Yes"],
    ["Low", "Good", "Yes", "Medium", "Yes"],
    ["High", "Good", "No", "High", "No"],
    ["Medium", "Poor", "No", "Low", "No"]
]

features = ["Income", "Credit", "Job", "Debt"]
target = "Loan"

positive = [row for row in data if row[-1] == "Yes"]

rules = []

while positive:

    best_rule = None
    best_covered = []

    for example in positive:

        for i in range(len(features)):

            rule = (features[i], example[i])

            covered = [
                row for row in positive
                if row[i] == example[i]
            ]

            if len(covered) > len(best_covered):
                best_rule = rule
                best_covered = covered

    if best_rule is None:
        break

    rules.append(best_rule)

    print("\nRule", len(rules))
    print(
        "IF",
        best_rule[0],
        "=",
        best_rule[1],
        "THEN Loan = Yes"
    )

    print("Covered positive examples:", len(best_covered))

    positive = [
        row for row in positive
        if row not in best_covered
    ]

print("\nFinal Rules")
print("-----------")

for i, rule in enumerate(rules, 1):
    print(
        "Rule", i,
        ": IF", rule[0], "=", rule[1],
        "THEN Loan = Yes"
    )
```
