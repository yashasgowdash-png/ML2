```python
data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"]
]

n = len(data[0]) - 1

S = ["Ø"] * n
G = [["?"] * n]

def covers(h, x):
    for i in range(n):
        if h[i] != "?" and h[i] != x[i]:
            return False
    return True

def more_general(h1, h2):
    for i in range(n):
        if h1[i] != "?" and h1[i] != h2[i]:
            return False
    return True

def generalize_S(s, x):
    new_s = s.copy()

    for i in range(n):
        if new_s[i] == "Ø":
            new_s[i] = x[i]
        elif new_s[i] != x[i]:
            new_s[i] = "?"

    return new_s

def specialize_g(g, x):
    result = []

    for i in range(n):
        if g[i] == "?":
            h = g.copy()
            h[i] = x[i]
            result.append(h)

    return result

for iteration, row in enumerate(data, 1):

    x = row[:-1]
    label = row[-1]

    print("\nIteration", iteration)
    print("Example:", x)
    print("Class :", label)

    if label == "Yes":

        S = generalize_S(S, x)

        new_G = []

        for g in G:
            if covers(g, x):
                new_G.append(g)

        G = new_G

    else:

        new_G = []

        for g in G:
            if covers(g, x):
                specializations = specialize_g(g, x)

                for h in specializations:
                    if more_general(h, S):
                        new_G.append(h)
            else:
                new_G.append(g)

        G = new_G

    print("S =", S)
    print("G =", G)

print("\nFinal Specific Boundary:")
print(S)

print("\nFinal General Boundary:")
for g in G:
    print(g)
```
